## 2: Lists and slicing

A list is ordered and zero-indexed: the first item is `[0]`. Slices are `[start:stop]`,
**including** start, **excluding** stop: `[:3]` is the first three, `[-1]` is the last
item, and `[-3:]` is the last three.

```python
scores = [88.5, 92.0, 79.5, 95.0, 84.0, 91.5]

print("first:", scores[0], "| last:", scores[-1])
print("first three:", scores[:3])
print("last three :", scores[-3:])

scores.append(77.0)          # modifies in place, returns nothing
scores.sort()                # also in place
print("sorted:", scores)
print("how many:", len(scores))
```

## 3: Dictionaries

A dict maps keys to values. It is the most load-bearing structure in the course: JSON
files, API responses, and every row a scraper produces are dicts or lists of dicts.

`d[k]` and `d.get(k)` behave differently on a missing key: the first raises an error, the
second returns `None` (or a supplied default). Both behaviours are useful, and Week 10
material requires choosing between them deliberately.

```python
ratings = {"Port City Java": 92.5, "Los Lobos": 88.0}

print(ratings["Port City Java"])          # crashes if absent
print(ratings.get("Talley Market"))       # None if absent
print(ratings.get("Talley Market", 0))    # default if absent

ratings["Talley Market"] = 90.5           # add or overwrite
print(len(ratings), "entries:", list(ratings.keys()))
```

```python
# Try it: look up "Los Lobos" with .get, then delete it with del, then look it up again.
print(ratings.get("Los Lobos"))
del ratings["Los Lobos"]
print(ratings)
ratings.update({"Los Lobos": 88.0})
print(ratings)
```

## 4: Loops

`for x in things:` walks any collection. Two patterns cover most data work: accumulate
(build a running total or list) and filter-count (tally the items meeting a condition).

```python
all_scores = [88.5, 92.0, 79.5, 95.0]

total = 0
n_low = 0
for s in all_scores:
    total = total + s
    if s < 90:
        n_low = n_low + 1

print(f"mean by hand: {total / len(all_scores):.2f}")
print(f"{n_low} of {len(all_scores)} scores below 90")

# looping over a dict gives keys; .items() gives both
for name, r in {"Port City Java": 92.5, "Los Lobos": 88.0}.items():
    print(f"  {name}: {r}")
```

```python
# Try it: count how many scores are 90 or above, and print it as a sentence.
count = 0
n_high = 0
for i in all_scores:
  count = count + i
  if i > 90:
    n_high = n_high + 1

print(f"{n_high} of {len(all_scores)} are above 90")
```

## 5: Comprehensions

The catalog lists comprehensions as a prerequisite. `[expression for x in things if
condition]` builds a list in one line, and it is the plain-Python shape of what `pandas`
does to whole columns. Read it left to right: *keep the expression, for each x, where the
condition holds.*

```python
raw = ["88.5", " 92.0", "79.5 ", "95.0"]

clean = [float(s) for s in raw]                 # transform every item
low   = [x for x in clean if x < 90]            # filter
names = {"Raleigh": 3, "Cary": 2, "Apex": 1}
big   = [city for city, n in names.items() if n >= 2]

print(clean)
print(low)
print(big)
```

```python
# Try it: one comprehension — the squares of 1 through 10, but only the even squares.
squares = ["1", "4", "9", "16", "25", "36", "49", "64", "81", "100"]
even = [int(x) for x in squares if int(x) % 2 ==0]
print(even)
```

## 6: Functions

`def` packages steps under a name. The docstring (the string directly under the `def`
line) states what the function expects; docstrings are graded from Week 9 onward.
Arguments can have defaults; `return` hands the value back.

```python
def mean(values):
    """Arithmetic mean of a non-empty list of numbers."""
    return sum(values) / len(values)

def share_below(values, cutoff=90):
    """Fraction of values strictly below cutoff."""
    hits = [v for v in values if v < cutoff]
    return len(hits) / len(values)

print(mean([2, 4, 9]))
print(share_below([88.5, 92.0, 79.5, 95.0]))          # default cutoff
print(share_below([88.5, 92.0, 79.5, 95.0], 95))      # your own cutoff
```

```python
# Try it: write spread(values) returning max minus min, and test it on two lists.

def spread(values):
  return max(values) - min(values)

print(f"difference of min and max number is: {spread([100, 2])}")
#print(spread([500, 3]))
```

## 7: None and truthiness

`None` is Python's "no value," distinct from zero, the empty string, and `False`. The
difference between `None` and `0` comes up constantly in data work. Test for it with
`is None`.

Empty values are "falsy": `[]`, `{}`, `""`, and `0` all count as `False` in an `if`. This
is convenient and occasionally dangerous: `if score:` treats a legitimate score of 0 the
same as a missing one.

```python
score = None
print(score is None)         # the right test
print(score == 0)            # False — None is not zero

for value in [0, "", [], None, 42, "0"]:
    print(f"{value!r:6} -> {'truthy' if value else 'falsy'}")

# note "0" (the string) is truthy. The string is not the number.
```

```python
# Try it: why might `if seats:` be the wrong test for "seats was recorded"?
# Write your answer as a comment, then demonstrate it with seats = 0.

# 'if seats' is the wrong test since it wont be able to tell the difference
# between None and 0. Python would treat the if seat = 0 as 'false' and if
# statement would not execute also it would treat a 0 as no data provided
# (what a None would do)
#. We should use if seats is not None... execute
# if block.

seats = 0
if seats is not None:
  #The if statement will evaluate since its True and not False
  print("Youve selected the right test since it will eval a 0")
```

## 8: Reading a traceback

Read a traceback **from the bottom up**: the last line names the exception and says what
went wrong; the lines above it point at the file and line. The cell below triggers an
error on purpose, catches it, and prints the parts.

```python
def total_seats(rows):
    total = 0
    for r in rows:
        total = total + r["seats"]
    return total

rows = [{"seats": 40}, {"seats": "55"}, {"seats": 32}]

try:
    total_seats(rows)
except TypeError as e:
    print("exception type   : TypeError")
    print("its message      :", e)
    print("what it means    : one of the seats values is a string, not a number.")
    print("where to look    : the line doing the +, and then at your DATA, not your code.")
```

## 9: Putting it together: a table, before pandas

Raw data usually arrives as a list of dicts, one dict per row, and the patterns above
compose to answer real questions about it. Working this example once by hand makes the
one-line `pandas` versions next week legible.

```python
inspections = [
    {"name": "Hargett Fish Camp",  "city": "Fuquay-Varina", "score": 86.7, "critical": 2},
    {"name": "Cabarrus Provisions","city": "Knightdale",    "score": 94.0, "critical": 0},
    {"name": "Jones Counter",      "city": "Knightdale",    "score": 95.2, "critical": 0},
    {"name": "Peace Smokehouse",   "city": "Holly Springs", "score": 100.0,"critical": 0},
]

scores = [r["score"] for r in inspections]
print(f"{len(inspections)} inspections, mean score {sum(scores)/len(scores):.1f}")

flagged = [r["name"] for r in inspections if r["critical"] > 0]
print("with critical violations:", flagged)

by_city = {}
for r in inspections:
    by_city[r["city"]] = by_city.get(r["city"], 0) + 1
print("inspections per city:", by_city)

# Next week, each of those is one line:
#df.scores.mean()
#df[df.critical > 0]
#df.city.value_counts()
```

```python
# Try it: mean score in Knightdale only. One comprehension plus the mean pattern.

#get all scores for city knightdale using a comprehension
kd_scores = [r["score"] for r in inspections if r["city"] == "Knightdale"]

#calculaate the mean score for all knightdale scores
kd_mean_score = sum(kd_scores) / len(kd_scores)

#print mean score with 1 digit past decimal
print(f"Mean score for Knightdale: {kd_mean_score:.1f}")
```
